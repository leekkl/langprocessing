```python
!pip install jupyter
import folium

# 1. Initialize the map centered on a specific coordinate
m = folium.Map(location=[40.7128, -74.0060], zoom_start=12, tiles="OpenStreetMap")

# 2. Add an interactive marker with a popup and tooltip
folium.Marker(
    location=[40.7128, -74.0060],
    popup="<b>Central Park</b><br>New York, NY",
    tooltip="Click for details",
    icon=folium.Icon(color="green", icon="info-sign")
).add_to(m)

# 3. Add a circle overlay to represent a radius
folium.Circle(
    location=[40.7128, -74.0060],
    radius=1000,  # Radius in meters
    color="blue",
    fill=True,
    fill_color="blue",
    fill_opacity=0.2
).add_to(m)

from ipyleaflet import Map, Marker, Circle

# Initialize map
m = Map(center=(40.7128, -74.0060), zoom=12)

# Create and add a draggable marker
marker = Marker(location=(40.7128, -74.0060), draggable=True)
m.add_layer(marker)

!pip freeze > requirements.txt
!pipreqsnb --savepath Path_to_save_the_requirements.txt Path_to_the_notebook.ipynb

# Any changes to the marker position in the notebook UI
# will instantly update `marker.location` in your Python environment!
```
