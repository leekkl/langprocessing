```python
!pip install shinywidgets
!pip install altair bokeh plotly ipyleaflet pydeck==0.8.0
from shiny import App, ui, render
import folium
import io
import nest_asyncio
from shiny import App, ui, run_app

# 1. Patch the existing Jupyter event loop
nest_asyncio.apply()

# 1. Define the User Interface (UI)
app_ui = ui.page_fluid(
    ui.h2("My Interactive Folium Map"),
    ui.p("This interactive map is hosted on Posit Connect Cloud."),
    # UI output element to display the raw HTML map
    ui.output_ui("map_output")
)

# 2. Define the Server Logic
def server(input, output, session):

    @render.ui
    def map_output():
        # Create a basic Folium Map centered on a specific location
        m = folium.Map(location=[40.7128, -74.0060], zoom_start=12)

        # Add a test marker
        folium.Marker(
            [40.7128, -74.0060],
            popup="New York City"
        ).add_to(m)

        # Save the map to an in-memory HTML buffer string
        data = io.BytesIO()
        m.save(data, close_file=False)
        map_html = data.getvalue().decode()

        # Return the raw map HTML safely to the Shiny layout
        return ui.HTML(map_html)

# 3. Initialize the App
app = App(app_ui, server)
```
